```python
import pandas as pd

female_path = "/Users/lyuxiaotian/Desktop/visualization/02_activities/assignments/ontario_top_baby_names_female.csv"  
male_path   = "/Users/lyuxiaotian/Desktop/visualization/02_activities/assignments/ontario_top_baby_names_male.csv"

df_f = pd.read_csv(female_path)
df_m = pd.read_csv(male_path)

rename_map = {
    "Year/Année": "year",
    "Name/Nom": "name",
    "Frequency/Fréquence": "freq",
    "Year": "year",
    "Name": "name",
    "Frequency": "freq",
}
df_f = df_f.rename(columns=rename_map)
df_m = df_m.rename(columns=rename_map)

df_f["gender"] = "Female"
df_m["gender"] = "Male"

df = pd.concat([df_f, df_m], ignore_index=True)

df["name"] = df["name"].astype(str).str.strip().str.upper()

top5_f = (
    df[df["gender"]=="Female"]
    .groupby("name")["freq"].sum()
    .nlargest(5)
)
top5_m = (
    df[df["gender"]=="Male"]
    .groupby("name")["freq"].sum()
    .nlargest(5)
)

with open("top5_names.txt", "w", encoding="utf-8") as f:
    f.write("Female Top 5:\n")
    f.write("\n".join(top5_f.index.tolist()))
    f.write("\n\nMale Top 5:\n")
    f.write("\n".join(top5_m.index.tolist()))

df_out = df[["year", "gender", "name", "freq"]].copy()
df_out.to_csv("names_clean.csv", index=False)

print("Saved: names_clean.csv and top5_names.txt")
print(top5_f)
print(top5_m)
```
