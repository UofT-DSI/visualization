```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(2)
x = np.arange(50)
y1=np.random.randint(0, 100, 50)
y2=np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = "lower left", frameon = True, fontsize = 10, ncol = 1, shadow = True)
fig.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x, y2, label = "Person 2")
ax.legend(loc = "lower left", bbox_to_anchor = (1,1))
fig.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
ax.text(20, 90, "This value is important!", ha = "center", color = "red", size = 10)
fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis ([0, 10, 0, 10])
ax.text (1, 5, ". Data: (1,5)", transform=ax.transData)
ax.text (0.5, 0.1, ". Axes: (0.5, 0.1)", transform=ax.transAxes)
ax.text (0.2, 0.2, ". Figure: (0.2, 0.2)", transform=fig.transFigure)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
ax.annotate('This is important!', xy = (10, 95), xytext=(20, 94), arrowprops=dict(arrowstyle='simple', color ="indigo"))
fig.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_locator(plt.NullLocator())
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
ax.xaxis.set_major_locator(plt.MultipleLocator(16))
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
plt.xticks(rotation = 45, ha = "right")
```

```python
font1 = {'family': 'serif', 'color': 'indigo'}
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower left")
plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```
