```python
#Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

Create ScatterPlot

```python
np.random.seed(613) #for repreducibility
x = np.arange(50)  #integers from 0 to 49
y = np.random.randint(0, 100, 50) #random integers between 0 and 100, 50 of them
# above is making fake data for the x and y axis


fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 (horizontal) by 3 (vertical)
ax.scatter(x,y)
```

Create Bar Chart

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.barh(x,y)

#horizontal bar chart, x and y are switched
```

Create Line Chart

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y)
```

Create Histogram

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.hist(y)
```

Line Chart with Title, X-Label and Y-Label

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout() #help with spacing of the title and axis labels, don't overlap with the figure, don't get cut off
```

Try Changing Font Style, Color and Size

```python
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict=font1, loc = 'left') #loc is location of the title, left, right, center (default, if not specified)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

Change Scatterplot's Marker Style

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.scatter(x,y,
           marker='*', #marker shape, default is 'o' for circle, '*' is star
           color='indigo') #color of the markers, default is 'blue')
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y, marker='*', color='indigo', linestyle='--', linewidth=2) #linestyle options: solid (default), dashed, dashdot, dotted 
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y, marker='*', markersize=12, color='#7425b9', linestyle='--', linewidth=2, markeredgecolor='#fa9359', markerfacecolor = '#000000') #markersize is size of the marker, default is 6, markeredgecolor is color of the edge of the marker, default is same as color, markeredgewidth is width of the edge of the marker, default is 1
```

```python
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y, marker='*', markersize=12, color='#7425b9', linestyle='--', linewidth=2, markeredgecolor='#fa9359', markerfacecolor = '#000000') #markersize is size of the marker, default is 6, markeredgecolor is color of the edge of the marker, default is same as color, markeredgewidth is width of the edge of the marker, default is 1
ax.grid(axis = 'y') #line come from y axis (horizontal one)
ax.grid(axis= 'x') #line come from x axis (vertical one)
```

```python
#Activity, change the grid line's look
fig, ax = plt.subplots(figsize=(5,3)) #size of the figure be 5 by 3
ax.plot(x,y, marker='*', markersize=12, color='#7425b9', linestyle='--', linewidth=2, markeredgecolor='#fa9359', markerfacecolor = '#000000') #markersize is size of the marker, default is 6, markeredgecolor is color of the edge of the marker, default is same as color, markeredgewidth is width of the edge of the marker, default is 1
ax.grid(axis = 'y', color='skyblue', linestyle='--', linewidth=2) #line come from y axis (horizontal one)
ax.grid(axis= 'x', color='pink', linestyle='--', linewidth=2) #line come from x axis (vertical one)
```
