```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, y2) #listing both y1 and y2 here also works
fig.show()
```

```python
# adding legends
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2, label='person 2')
ax.legend(loc='lower right')
fig.show()
```

```python
# adding legends
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2, label='person 2')
ax.legend(loc='best') # can use best for it to decide
fig.show()
```

```python
# modifying the legend
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2, label='person 2')
ax.legend(loc='lower right',
          frameon = True, # frame 
          fontsize = 12,
          ncol = 2, # changing the number of columns 
          shadow = True)
fig.show()
```

```python
# moving the legend outside 
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='person 1')
ax.plot(x,y2, label='person 2')
ax.legend(loc='upper left',
          bbox_to_anchor = (1,1))
fig.show()
```

```python
# adding text and annotations
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower left')
ax.text(10,95,'this value is important!',# the values correspond with the coordinates for the axis
ha ='center', #ha = horizontal alignment
color = 'red',
size=12) 
fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5,'.Data:(1,5)', transform=ax.transData)
ax.text(0.5,0.1,'.Data:(0.5,0.1)', transform=ax.transAxes) # in this case, the placement is relative to the whole graph, not corresponding to the axis cooredinate)
ax.text(0.5,0.1,'.Figure:(0.2,0.2)', transform=fig.transFigure) # in this case, it will be relative to the whole figure now 
```

```python
#adding arrows
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')
ax.annotate('This is important', xy=(10,95), xytext=(20,94),
             arrowprops=dict(facecolor='black'))
            
fig.show()
```

```python
#adding diff arrows arrows
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')
ax.annotate('This is important', xy=(10,95), xytext=(20,94),
             arrowprops=dict(facecolor='black'))
ax.annotate('This is also important', xy=(10,10), xytext=(20,9),
             arrowprops=dict(arrowstyle='wedge',color='hotpink'))
            
fig.show()
```

```python
#locators = remove ticks 
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) # this removes x axis ticks
```

```python
#formatters = remove just numbers
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_formatter(plt.NullFormatter()) # just removes the numbers but keeps the ticks
```

```python
#setting the number of ticks
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3)) 
```

```python
#change the interval length of each ticks
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.MultipleLocator(5)) 
```

```python
#rotating the axis 
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label='person 1')
ax.scatter(x,y2, label='person 2')
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
```

```python
#making wordclouds 

from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')
df
```

```python
text=' '.join(each for each in df.quote)

wordcloud=WordCloud(background_color="white", colormap='inferno').generate(text)

# use matplotlib to make the image 
fig,ax =plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
```

```python
#making venn diagram

from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A=set(['apple', 'banana', 'watermelon'])
B=set(['pumpkin', 'blueberry', 'apple', 'key lime'])
```

```python
diagram=venn2_unweighted([A,B], 
                         set_labels=('fruits', 'pies'),
                         set_colors=('blue', 'red'),
                         alpha=0.5)
# defining rules so things correctly appear in the venn diagrams
diagram.get_label_by_id('10').set_text("\n".join(A-B))
diagram.get_label_by_id('11').set_text("\n".join(A&B))
diagram.get_label_by_id('01').set_text("\n".join(B-A))
```
