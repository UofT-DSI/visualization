Subplots

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL   #library for image processing, called Pillow
import requests

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y2 = np.array([110, 180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, #number of columns
                               nrows = 1, #number of rows
                               figsize = (7,3)) #size of the figure
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

Activity_Customize the Plot

```python
# change scatter plot with * and in pale pink color, rotate x axis labels by 45 degrees, change bar plot to have light blue bars with black edges
fig, (ax1, ax2) = plt.subplots(ncols=2, #number of columns
                               nrows = 1, #number of rows
                               figsize = (7,3)) #size of the figure
ax1.scatter(x1,y1, marker='*', color='pink')
ax1.set_xticklabels(x1, rotation=45)
ax2.bar(x2,y2, color='lightblue', edgecolor='black')
ax2.set_xticklabels(x2, rotation=45)
```

Demonstration

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, #number of columns
                               nrows = 1, #number of rows
                               figsize = (9,3)) #size of the figure

font1 = {'family': 'serif', 'size': 10, 'color': 'indigo'}
font2 = {'family': 'serif', 'size': 14, 'color': 'darkblue'}


ax1.scatter(x1,y1, marker='o', color='red')
ax2.bar(x2,y2, color='darkgreen')

ax1.set_title("Scatter Plot", fontdict=font2)
#ax2.set_title("Bar Plot", fontdict=font2)
ax1.set_xlabel("Time", fontdict=font1)
ax1.set_ylabel("Measure", fontdict=font1)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7,4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1, color='purple')
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7,4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1, color='purple')

someaxes["ax1"].set_xlabel("A Big Label", fontsize = 18)
someaxes["ax2"].set_xlabel("Another Label", fontsize = 18)
someaxes["ax3"].set_xlabel("Label 2: 2 Fast 2 Furious", fontsize = 18)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7,4),
                                   layout='constrained') #constrained layout automatically adjusts the spacing between subplots to prevent overlap
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1, color='purple')

someaxes["ax1"].set_xlabel("A Big Label", fontsize = 18)
someaxes["ax2"].set_xlabel("Another Label", fontsize = 18)
someaxes["ax3"].set_xlabel("Label 2: 2 Fast 2 Furious", fontsize = 18)
```

Add Error information

```python
y2_sd = np.std(y2) #standard deviation of y2, which is a measure of how spread out the values in y2 are from the mean of y2
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')
```

```python
y2_sd = np.std(y2) 
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x, #our x values
            y2, #our y values
            yerr=y2_sd, 
            fmt="none" #, format of the error bars, try to distinct from the original color, "none" means no marker will be used for the data points
            #, color='red', ecolor='lightgray', elinewidth=3, capsize=0
            )
```

```python
y2_sd = np.std(y2) 
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x, #our x values
            y2, #our y values
            yerr=y2_sd, 
            fmt="none" #, format of the error bars, try to distinct from the original color, "none" means no marker will be used for the data points
            , ecolor='indigo',elinewidth = 4, capsize = 6, capthick = 4
            )
```

```python
y2_sd = np.std(y2) 
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax.errorbar(x, #our x values
            y2, #our y values
            yerr=y2_sd, 
            fmt="none" 
            , ecolor='indigo',elinewidth = 4, capsize = 6, capthick = 4,
            errorevery = 2 #, display error bars on every 2nd data point (not commonly use)
            )
```

Add Image to the Plot

```python
from PIL import Image
import requests
from io import BytesIO
```

```python
response = requests.get('https://static.wikia.nocookie.net/the-ultimate-good/images/3/3d/Luffy.png/revision/latest?cb=20230417200546')
img_file = BytesIO(response.content) #create a file-like object from the response content, which is the image data
image = Image.open(img_file) #open the image file using PIL
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')

ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35]) 

ax_image.imshow(image) #display the image on the new axes, which is a small subplot within the original plot, located at (0.1, 0.11) with a width of 0.15 and a height of 0.35 relative to the original plot
ax_image.axis('off') #turn off the axis for the image
```
