```python
# Social Housing Waiting List Visualization, Zahra Sepahi
# City of Toronto Open Data Portal
# Link to complete dataset: https://open.toronto.ca/dataset/centralized-waiting-list-activity-for-social-housing/

import pandas as pd
import matplotlib.pyplot as plt

# Download the dataset
df = pd.read_csv("/Users/zahrasepahi/visualization/02_activities/assignments/centralized_waiting_list_activity.csv")

# Evaluate only the relevant columns
df = df[["Quarter/Year", "Total active waiting list"]]

# Remove rows with missing values
df = df.dropna()

# Convert Quarter/Year to string for plotting
df["Quarter/Year"] = df["Quarter/Year"].astype(str)

# Visualization
plt.figure(figsize=(10, 6))
plt.plot(df["Quarter/Year"], df["Total active waiting list"], marker='o', color='teal', linewidth=2)
plt.xlabel("Quarter / Year")
plt.ylabel("Total Active Waiting List (Households)")
plt.title("Toronto Social Housing Waiting List Over Time")
plt.xticks(df["Quarter/Year"][::2], rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Save the figure for submission
plt.savefig("/Users/zahrasepahi/visualization/02_activities/assignments/toronto_social_housing_waitlist.png")
plt.show()

# My apologies for this average looking plot. The trend line doesn't seem to make sense and didn't know how to fix it. :( 
```
