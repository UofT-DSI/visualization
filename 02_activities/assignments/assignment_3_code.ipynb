The raw data was obtained from: https://open.toronto.ca/dataset/deaths-of-shelter-residents/ 

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
df = pd.read_csv("deaths_residents.csv") #Reads our csv file data.
print(df)
```

```python
graph = sns.lineplot(data=df, x="Year", y="Deaths", hue = "Sex", palette = ["royalblue", "hotpink"]) #Using sns.lineplot we can easily split our data to male and female and have 2 lines on one graph. 
graph.set_xticks(sorted(df["Year"].unique())) #Ensures that the program doesn't automatically add decimals to our x-axis data (i.e., years).
graph.tick_params(axis = "x", rotation = 45) #Allows us to show the years on x-axis without overlap as there are many of them. 
graph.set(title = "Number of decedents in shelters", xlabel = "Year", ylabel = "Decedents") #Adds appropriate labels. 
```
