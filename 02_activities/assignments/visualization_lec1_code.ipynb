```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(5)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
font1 = {'family': 'sans-serif', 'color': 'blue', 'size':20}
font2 = {'family': 'monospace', 'color': 'pink', 'size': 10}
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = "right")
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x, y, marker = '*', color = 'indigo', linestyle = '--', linewidth = 3)
fig.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x, y, marker = '*',markersize = 12,  color = '#7425b9', linestyle = '--', 
        linewidth = 1.5, markeredgecolor = '#fa9359', 
        markerfacecolor = '#000000' )
fig.show()
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.grid(axis = 'y', color = 'skyblue', linewidth =2, linestyle = '-.')
ax.plot(x, y, marker = '*',markersize = 12,  color = '#7425b9', linestyle = '--', 
        linewidth = 1.5, markeredgecolor = '#fa9359', 
        markerfacecolor = '#000000' )
```
